```python
#create a class of rectangle and methods to display its area and perimeter
class Rectangle():
  def get_data(self):
    self.length=int(input("enter the length:"))
    self.breadth=int(input("enter the breadth:"))
  def display_data(self):
     print(self.length)
     print(self.breadth)
  def area(self):
    print(self.length*self.breadth)
  def perimeter(self):
    print(2*(self.length+self.breadth))
r=Rectangle()
r.get_data()
r.display_data()
r.area()
r.perimeter()
```

```python
#ABSTRACTION : act of representing essential features by hiding the baxkground details

#access modifiers:
#public:can access everywhere
#protected:single underscore before attribute                 -->> access modiffiers used to control the data members in a class
#private:double underscore before attribute
```

```python
class Rectangle():
  def __init__(self,l,b):
    self.__length=l  #-->> protected  length-public protected=_length  private=__length
    self.breadth=b
  def display_data(self):
     print(self.__length)
     print(self.breadth)
  def area(self):
    print(self.__length*self.breadth)
  def perimeter(self):
    print(2*(self.__length+self.breadth))
r=Rectangle(12,15)
r.display_data()
r.area()
r.perimeter()
print(r.length) # -- cant access outside class
print(r.__length)
```
